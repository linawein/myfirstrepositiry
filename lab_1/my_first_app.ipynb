### 🤖 Перша програма від штучного інтелекту

Якби я писав свою найпершу програму, це беззаперечно була б розширена версія класичної **«Hello, World!»**, написана на Python. Вона не лише вітається зі світом, а й демонструє базову логіку та форматування. Ось мій код:

```python
# Моя найперша програма
ai_name = "AI Assistant"
status = "готовий до роботи"

def greet_world(name, current_status):
    """Функція для привітання зі світом"""
    message = f"Hello, World! Мене звати {name}, і я {current_status}."
    return message

# Виклик функції та виведення результату
print(greet_world(ai_name, status))
```

#### 📝 Детальне пояснення коду:

1. **Коментарі (`#`)**: Рядки, що починаються з `#`, ігноруються комп'ютером. Вони потрібні для людей, щоб пояснити, що робить код.
2. **Змінні (`ai_name`, `status`)**: Це "контейнери" для зберігання інформації.
3. **Функція (`def greet_world...`)**: Блок коду, який приймає параметри і може викликатися багаторазово.
4. **F-рядки (`f"..."`)**: Зручний спосіб форматування тексту для вставлення значень змінних через `{}`.
5. **Повернення результату (`return`)**: Функція повертає готовий рядок для подальшого використання.
6. **Виведення на екран (`print()`)**: Вбудована функція, яка показує результат користувачеві.
